# Basic Data Pre-processing

**Input:** `Solar_Energy.csv` (raw) → **Output:** `Cleaned_Dataset.csv` (used by the EDA and model-building notebooks)

```python
# Imports
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

### 1. Load Data

```python
# Load Original Dataset
data = pd.read_csv(r"D:\OJT\Solar_Energy_Production_Analysis\data\Solar_Energy.csv")
data.head()
```

### 2. Create a Copy of the Data

`data.copy()` makes an independent (deep) copy, so changes made to `df` do not affect the original `data`.

```python
df = data.copy()
```

### 3. Basic Understanding

Check the dimensions, column names, column types (numeric vs categorical) and data types.

```python
print("Number of rows and columns:\n", df.shape)
print("\n Column names:\n", df.columns.to_list())
print("\n Categorical Columns: - \n", df.select_dtypes(include = "object").columns.to_list())
print("\n Numeric Column: - \n", df.select_dtypes(include = "number").columns.to_list())
```

```python
# Information about the data
df.info()
```

```python
# Check data types
df.dtypes
```

**Observations: data types that need correction**
- `Interconnection Date` (str) → `datetime`.
- `Zip` (float) → string, because it is an identifier, not a quantity (leading zeros must be preserved).
- `Data Through Date` (str) is **not** converted because the column holds a single value and is dropped in section 8.

### 4. Summary Statistics

```python
df.describe()
```

**Observations**
- Numeric columns are extremely right-skewed: median `Estimated PV System Size (kWdc)` is 7.13 but the maximum is 43,470; median annual production is 8,364 kWh but the maximum is about 51 million kWh.
- `Number of Projects` is 1 in every row (std = 0), so it carries no information.

### 5. Check Duplicate Rows and Remove Them

```python
print("Number of duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()
```

**Observation:** No duplicate rows were found, so nothing was removed.

### 6. Check Null Values

```python
null_df = pd.DataFrame({
    "null_count": df.isnull().sum(),
    "null_%": (df.isnull().mean() * 100).round(2)
})
null_df[null_df["null_count"] > 0].sort_values("null_count", ascending=False)
```

### 7. Count Unique Values

```python
print("Number of unique values in each column:\n")
df.nunique()
```

```python
print("Repeated Project IDs:", df["Project ID"].duplicated().sum(), "of", len(df))
```

**Observations**
- **Project ID:** 218,077 unique values among 218,112 non-null rows → it is an identifier, not a feature. 35 IDs are repeated; there are no fully duplicate rows (section 6), so these rows differ in at least one other column. This should be checked (genuine re-submissions vs. data-entry errors).
- **Data Through Date** and **Number of Projects:** only 1 unique value each → no information, dropped in section 8.
- **High-cardinality categoricals:** `Circuit ID` (6,429), `Developer` (2,567), `Substation` (2,255), `City/Town` (2,965), `Zip` (1,729) and `Division` (109). `County` has 62 values (moderate).
- One-hot encoding the high-cardinality columns would create thousands of sparse columns, leading to heavy memory use and overfitting. They are therefore encoded with a **TargetEncoder** in the model notebook, and `Circuit ID` is dropped (section 8).

**Observations**
- `Data Through Date` and `Number of Projects` each have only 1 unique value, so they carry no information and can be dropped.
- `Division`, 'Developer', 'Metering Method', 'Energy Storage System Size (kWac)' and `Substation` handel null using pipeline
- `Project ID` is an identifier. It is kept for reference only and is not used as a model feature.

### 8. Drop Unnecessary Columns

- **Data Through Date** Single constant value (no information)
- **Number of Projects** Always 1 (no information)
- **Circuit ID** Very high cardinality (6,429 values) with little generalisable signal
- **Project ID** is **kept** for future reference (it is not a model feature).

```python
drop_cols = ["Data Through Date", "Number of Projects", "Circuit ID"]
df = df.drop(columns = drop_cols)
```

### 9. Handle Null Values (Drop Rows)

`Project ID`, `Zip` and `City/Town` have only a handful of missing values (< 0.05%), so those rows are dropped.

```python
# Very few gaps in `Project ID`, `Zip` and `City/Town` -> drop these rows
df = df.dropna(subset = ["Project ID", "Zip", "City/Town"]).reset_index(drop=True)
```

### 10. Fix Inconsistent City/Town Values

The same place is written in different cases (e.g. `STATEN ISLAND` 6,884 rows vs `Staten Island` 5,707 rows; `BROOKLYN` vs `Brooklyn`). Strip extra spaces and convert to title case so they merge into one category.

```python
# Strip extra spaces, capitalize properly
df['City/Town'] = df['City/Town'].astype(str).str.strip().str.title()
```

### 11. Final Check

Verify the shape, remaining null values, duplicates and final data types.

```python
print("Shape:", df.shape)
print("Total null values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
df.dtypes
```

```python
path = r"D:\OJT\Solar_Energy_Production_Analysis\data\Cleaned_Dataset.csv"
df.to_csv(path, index = False)
print("Saved:", df.shape)
print("\n1st five rows\n")
df.head()
```
